```python
import ollama
import pandas as pd

print("Python notebook is working!")
print("Ollama and Pandas imported successfully!")
```

```python
response = ollama.chat(
    model="llama3.2:1b",
    messages=[
        {
            "role": "system",
            "content": (
                "You are an Education Assistant. "
                "Answer in simple English with clear examples. "
                "If you are unsure, say so."
            )
        },
        {
            "role": "user",
            "content": "What is DBMS?"
        }
    ]
)

print("Question: What is DBMS?")
print("Answer:", response["message"]["content"])
```

```python
import pandas as pd
from datetime import datetime
from pathlib import Path

Path("dataset").mkdir(exist_ok=True)

result = pd.DataFrame([{
    "timestamp": datetime.now().isoformat(timespec="seconds"),
    "question": "What is DBMS?",
    "response": response["message"]["content"]
}])

result.to_csv(
    "dataset/responses.csv",
    index=False
)

print("Test result saved successfully!")
```
